```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
data = pd.read_csv("datasets/water_potability.csv")
data.head()
```

```python
data.isnull().sum()
```

```python
data = data.dropna()
```

```python
data.isnull().sum()
```

```python
plt.figure(figsize=(15, 10))
sns.countplot(
    x="Potability",
    data=data,
    hue="Potability",
    palette={0: "lightblue", 1: "orange"},
    legend=False
)
plt.title("Distribution of unsafe and safe water")
plt.xlabel("Potability (0 = Unsafe, 1 = Safe)")
plt.ylabel("Count")
plt.show()
```
